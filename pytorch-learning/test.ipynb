```python
#Testing speed between CPU and GPU
import time
size = 10000

#create random matrices on CPU
matrixcpu1 = torch.rand(size, size)
matrixcpu2 = torch.rand(size, size)

#measure time on CPU
start_time = time.time()
resultcpu = torch.matmul(matrixcpu1, matrixcpu2)
end_timecpu = time.time() - start_time

print(f"Time taken on CPU: {end_time:.4f} seconds")

#create random matrices on GPU
matrixgpu1 = matrixcpu1.to(device)
matrixgpu2 = matrixcpu2.to(device)

#measure time on GPU
start_time = time.time()
resultgpu = torch.matmul(matrixgpu1, matrixgpu2)
end_timegpu = time.time() - start_time

print(f"Time taken on GPU: {end_time:.4f} seconds")

#Compare results
print(f"/nSpeedUP (GPU Time / CPU Time): {end_timecpu / end_timegpu:.2f}x")    
```

```python

import torch
import time

# Check MPS availability
if not torch.backends.mps.is_available():
    raise RuntimeError("MPS GPU is not available")

device_cpu = torch.device("cpu")
device_gpu = torch.device("mps")

# Matrix size
N = 1024
ITERATIONS = 50
WARMUP = 5

# Create matrices on respective devices
a_cpu = torch.randn(N, N, device=device_cpu)
b_cpu = torch.randn(N, N, device=device_cpu)

a_gpu = torch.randn(N, N, device=device_gpu)
b_gpu = torch.randn(N, N, device=device_gpu)

def benchmark(a, b, device):
    # Warm-up
    for _ in range(WARMUP):
        result = a @ b

    if device == "mps":
        torch.mps.synchronize()

    # Start timing
    start = time.perf_counter()

    for _ in range(ITERATIONS):
        result = a @ b

    # Wait until GPU finishes
    if device == "mps":
        torch.mps.synchronize()

    end = time.perf_counter()

    return (end - start) * 1000

# Run benchmarks
cpu_time = benchmark(a_cpu, b_cpu, "cpu")
gpu_time = benchmark(a_gpu, b_gpu, "mps")

print(f"CPU Time: {cpu_time:.2f} ms")
print(f"GPU Time: {gpu_time:.2f} ms")

print(f"Speedup: {cpu_time / gpu_time:.2f}x")
```

##Implement gradient descent from scratch using Python

```python
#Initial weight 
weight = 0.0

#Learning rate 
learning_rate = 0.1

#Training iterations 
epochs = 20

for epoch in range(epochs):
    #calculate loss 
    loss = (weight - 3) ** 2
    
    #calculate gradient manually 
    gradient = 2 * (weight -3)
    
    #Update weight 
    weight = weight - learning_rate * gradient
    
    print(
        f"epoch: {epoch + 1}, "
        f"weight: {weight:.4f}, "
        f"Loss: {loss:.4f}"
    )

print("Final weight: ", weight)
```

##Recap of whole learning with y = 2x + 1

```python
import torch
import torch.nn as nn

#Reproducible synthetic data: y = 2x + 1
torch.manual_seed(42)
x = torch.randn(100, 1)
y = 2.0 * x + 1.0

#Define a simple linear model (1 input feature -> 1 output)
model = nn.Linear(in_features=1, out_features=1)

#Choose loss function and optimizer 
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)

#Training loop
for epoch in range(100):
    #Forward Pass: Make Predictions
    preds = model(x)
    loss = criterion(preds, y)
    
    #Backward pass: clear old gradients, compute new ones, uodate weights
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
print(f"Learned weight: {model.weight.item():.4f} (target: 2.000)")
print(f"Learned bias: {model.bias.item():.4f} (target: 1.000)")
```
