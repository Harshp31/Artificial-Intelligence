###Building a tiny neural network that predicts whether a student will get placed based on their CGPA.

```python
import torch

#Input data 
x = torch.tensor([
    [5.0],
    [6.0],
    [7.0],
    [8.0],
    [9.0]
])

#Actual placements results
y = torch.tensor([
    [0.0],
    [0.0],
    [1.0],
    [1.0],
    [1.0]
])
```

```python
#initialize weight and bias 

#Model parameters
weight = torch.tensor([[1.0]], requires_grad=True)
bias = torch.tensor([0.0], requires_grad=True)
```

```python
#Forward Propagation

#calculate weighted input
z = x * weight + bias

#Apply sigmoid activation 
prediction = torch.sigmoid(z)

print(prediction)
```

```python
#For Binary classification, we commonly use Binary Cross Entropy(BCE)

loss_fn = torch.nn.BCELoss()

loss = loss_fn(prediction, y)

print("Loss:", loss)
```
