###Building a tiny neural network that predicts whether a student will get placed based on their CGPA.

```python
import torch

#Input data 
x = torch.tensor([
    [5.0],
    [6.0],
    [7.0],
    [8.0],
    [9.0]
])

#Actual placements results
y = torch.tensor([
    [0.0],
    [0.0],
    [1.0],
    [1.0],
    [1.0]
])


```

```python
#initialize weight and bias 

#Model parameters
weight = torch.tensor([[1.0]], requires_grad=True)
bias = torch.tensor([0.0], requires_grad=True)
```

```python
#Forward Propagation

#calculate weighted input
z = x * weight + bias

#Apply sigmoid activation 
prediction = torch.sigmoid(z)

print(prediction)
```

```python
#For Binary classification, we commonly use Binary Cross Entropy(BCE)

loss_fn = torch.nn.BCELoss()

loss = loss_fn(prediction, y)

print("Loss:", loss)
```

```python
#Autograd 
loss.backward()

print("weighted gradient", weight.grad)
print("Bias gradient", bias.grad)
```

```python
#Input data 
x = torch.tensor([
    [5.0],
    [6.0],
    [7.0],
    [8.0],
    [9.0]
])

#Actual placements results
y = torch.tensor([
    [0.0],
    [0.0],
    [1.0],
    [1.0],
    [1.0]
])

#Normalize CGPA
X = x / 10.0

#Initialize trainable parameters
weight = torch.nn.Parameter(torch.tensor([[1.0]]))
bias = torch.nn.Parameter(torch.tensor([0.0]))

#Loss Function 
loss_fn = torch.nn.BCELoss()

#Optimizer 
optimizer = torch.optim.Adam([weight, bias], lr=0.1)


#Training loop
for epoch in range(500):
    #1. Forward Propogation 
    z = X * weight + bias 
    prediction = torch.sigmoid(z)

    #calculate loss
    loss = loss_fn(prediction, y)

    #clear previous gradients
    optimizer.zero_grad()

    #BackPropagation 
    loss.backward()

    #update parameters
    optimizer.step()



if(epoch + 1) % 100 == 0:
    print(
        f"epoch: {epoch + 1 }, "
        f"loss: {loss.item():.4f}"
    )
    
print("\nFinal Weight: ", weight.item())
print("Final Bias: ", bias.item())
```

```python
#Testing our trained neuron

# New student with CGPA 7.5
new_cgpa = torch.tensor([[7.5 / 10]])

# We don't need gradients during prediction
with torch.no_grad():

    z = new_cgpa * weight + bias
    probability = torch.sigmoid(z)

    print("Placement probability:",
          probability.item())

    if probability.item() >= 0.5:
        print("Prediction: Placed")
    else:
        print("Prediction: Not Placed")
```
