## Broadcasting


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array([1, 2, 3])
b = 2
print(a + b)
```

## Scalar Broadcasting

```python
a = np.array([1, 2, 3, 4])
print(a * 10)
```

## 2D-Broadcasting

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
b = np.array([10, 20, 30])

print(a + b)
```

## Column Broadcasting

```python
a = np.array([[1, 2, 3, 4],
              [5, 6, 7, 8]])

b = np.array([[10],
              [20]])

print(a + b)
```

```python
array1 = np.array([[1, 2, 3], [4, 5, 6]])
array2 = np.random.rand(3, 3)
array3 = np.zeros((4, 4))
```
