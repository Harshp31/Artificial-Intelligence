## Numpy Array And Basics

```python
import numpy as np
```

### Creating array from list


```python
arr_1d = np.array([1, 2, 3, 4, 5])
print("1D Array: ", arr_1d)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("2D array: ", arr_2d)
```

### List vs Numpy


```python
py_list = [1, 2, 3]
print("Python list Multiplication: ", py_list * 2)

np_array = np.array([1, 2, 3])
print("Python array Multiplication", np_array * 2)

import time

start = time.time()
py_list = [i*2 for i in range (100000)]
print("\n List Operation time: ", time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("\n Numpy Operation time: ", time.time() - start)
```

## Creating array from scratch


### Special arrays


```python
zeroes = np.zeros((3, 4))
print("Zeroes arry: \n", zeroes)

ones = np.ones((2, 3))
print("Ones array: \n", ones)

identity = np.eye(4)
print("Identity Matrix: \n", identity)

full = np.full((3, 3), 7)
print("Full of 7 array: \n", full)
```

### Ranges


```python
arrange = np.arange(0, 20, 5)    #20 is non inclusive outer boundary
print(f"Counting from 0 to twenty with multiples of 5: {arrange} ")

divide = np.linspace(0, 1, 5)
print(f"Divide 0 to 1 in 5 parts: {divide}")
```

## Random (Essentials in ML)

```python
random = np.random.random((2,3))
print(f"random array: \n {random}")

see = np.random.seed(42)
print(see)

#Create random array of 10 elements

elements = np.random.randn(10)
print(elements)

#create 4*4 random matrix
matrix = np.random.randint(1, 20, (4, 4))
print(matrix)

#Shuffle an array
element = np.array([1, 2, 4, 5, 6])
np.random.shuffle(element)
print(element)

#Pick 3 Random Numbers
np.random.choice(element, 3)
```
